9. 다음 두 배열 a, b를 하나의 1차원 배열로 연결하시오.

결과가 다음과 같이 나오도록 하시오.

[10 20 30 40 50 60]


```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([40, 50, 60])

x = np.concatenate((a,b), axis=0)
print(x)

```

10. 다음 1차원 배열 a를 **2행 3열**의 2차원 배열로 변경하시오.

[1 2 3 4 5 6]


```python
import numpy as np

a = np.array([1, 2, 3, 4, 5, 6])

b = a.reshape(2,3)
print(b)
```

11. 다음 배열 a를 `reshape()`와 `-1`을 사용하여 **3열을 가지는 2차원 배열**로 변경하시오.


```python
import numpy as np

a = np.arange(1, 13)

b = a.reshape(-1,3)
print(b)
```

12. 다음 1차원 배열 a의 shape은 `(4,)`이다.

newaxis를 사용하여 배열을 **4행 1열** 형태로 바꾸고, 변경된 배열과 shape을 출력하시오.


```python
import numpy as np

a = np.array([10, 20, 30, 40])

b = a.reshape(4,1)
print(b)
print(b.shape)
```

13. 다음 1차원 배열 a를 newaxis를 사용하여 **1행 4열** 형태로 바꾸고, 변경된 배열과 shape을 출력하시오.


```python
import numpy as np

a = np.array([10, 20, 30, 40])

b = a[np.newaxis,:]
print(b)
print(b.shape)
```
